```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots()

# Predicted most visited countries in 2025 and tourist counts (millions)
Countries = ['France', 'Spain', 'USA', 'Turkey']
counts = [100, 89, 79, 60]
bar_colors = ['tab:blue', 'tab:orange', 'tab:green', 'tab:red']

# Creating a horizontal bar chart
ax.barh(Countries, counts, color=bar_colors)
ax.invert_yaxis()  # Most visited at the top

ax.set_xlabel('Number of Tourists (Millions)')
ax.set_title('Most Visited Countries in 2025 (Estimated)')

plt.show()
```

```python
import matplotlib.pyplot as plt

labels = ['Services', 'Industry', 'Tourism', 'Agriculture']
sizes = [55, 22, 17, 6]

explode = (0.1, 0, 0, 0)

fig, ax = plt.subplots()
ax.pie(sizes, explode=explode, labels=labels, autopct='%1.1f%%',
       shadow=True, startangle=90, colors=['#3498db', '#2ecc71', '#e74c3c', '#f1c40f'])

ax.set_title("Turkey's Revenue Share by Sector (Estimated)", fontsize=14, fontweight='bold')
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

countries = ("United States", "China", "Japan", "Australia", "France")

medal_data = {
    'Bronze': (42, 24, 13, 16, 22),
    'Silver': (44, 27, 12, 19, 26),
    'Gold': (40, 40, 20, 18, 16),
    'Total': (126, 91, 45, 53, 64)
}

x = np.arange(len(countries))
width = 0.20
multiplier = 0

fig, ax = plt.subplots(figsize=(10, 6), layout='constrained')

for medal_type, counts in medal_data.items():
    offset = width * multiplier
    rects = ax.bar(x + offset, counts, width, label=medal_type)
    ax.bar_label(rects, padding=3, fontsize=9)
    multiplier += 1

ax.set_ylabel('Number of Medals')
ax.set_title('Top 5 Nations by Medals (Paris 2024)')
ax.set_xticks(x + width * 1.5)
ax.set_xticklabels(countries)
ax.legend(loc='upper right')
ax.set_ylim(0, 140)

plt.show()
```
